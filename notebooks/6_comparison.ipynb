```python
import json as _json

exp_dirs = {
    'Baseline': ROOT / 'outputs' / 'swinv2_small_baseline' / 'baseline_results.json',
    'Exp1 WRS+Focal':   ROOT / 'outputs' / 'swinv2_small_exp1'  / 'exp1_results.json',
    'Exp2 +Aug':        ROOT / 'outputs' / 'swinv2_small_exp2'  / 'exp2_results.json',
    'Exp3 2-Phase':     ROOT / 'outputs' / 'swinv2_small_exp3'  / 'exp3_results.json',
    'Exp5 Downsample':  ROOT / 'outputs' / 'swinv2_small_exp5'  / 'exp5_results.json',
}

all_results = {}
for label, path in exp_dirs.items():
    try:
        with open(path) as f: all_results[label] = _json.load(f)
        print(f'  [✓] {label}')
    except FileNotFoundError:
        print(f'  [–] {label}  (not yet run)')

print()
W = 95
print('='*W)
print(f'  {"Metric":<20}', end='')
for label in all_results: print(f'  {label:>13}', end='')
print()
print('='*W)

for metric_name, key, fmt in [
    ('Test QWK',   'qwk',      '.4f'),
    ('Macro F1',   'macro_f1', '.4f'),
]:
    print(f'  {metric_name:<20}', end='')
    for r in all_results.values():
        print(f'  {r[key]:>13{fmt}}', end='')
    print()

print()
print(f'  {"Per-class F1":<20}')
for g, name in enumerate(CLASS_NAMES):
    print(f'    {name:<20}', end='')
    for r in all_results.values():
        print(f'  {r["f1_per"][g]:>13.4f}', end='')
    print()
print('='*W)
```
