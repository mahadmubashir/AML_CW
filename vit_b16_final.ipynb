### Class Weights & Sampler

Compute balanced inverse-frequency weights from the training labels. Used by `WeightedRandomSampler` to rebalance batch composition during training.

```python
class DRDataset(Dataset):
    def __init__(self, df, data_dir, transform=None):
        self.df        = df.reset_index(drop=True)
        self.data_dir  = Path(data_dir)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row   = self.df.iloc[idx]
        image = Image.open(self.data_dir / f"{row['image']}.jpeg").convert('RGB')
        label = int(row['level'])
        if self.transform:
            image = self.transform(image)
        return image, label


# ViT requires symmetric [−1, 1] normalisation, not ImageNet RGB stats
train_transform = transforms.Compose([
    transforms.Resize(int(IMAGE_SIZE * 1.05)),
    transforms.RandomCrop(IMAGE_SIZE),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomApply([transforms.RandomRotation(degrees=5)], p=0.2),
    transforms.RandomApply([transforms.ColorJitter(brightness=0.05, contrast=0.05)], p=0.15),
    transforms.RandomApply([transforms.RandomAffine(degrees=0, translate=(0.02, 0.02))], p=0.15),
    transforms.ToTensor(),
    transforms.Normalize(VIT_MEAN, VIT_STD),
])

val_transform = transforms.Compose([
    transforms.Resize(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(VIT_MEAN, VIT_STD),
])

train_ds = DRDataset(train_df, TRAIN_DIR, transform=train_transform)
val_ds   = DRDataset(val_df,   VAL_DIR,   transform=val_transform)
test_ds  = DRDataset(test_df,  TEST_LABELS_DIR, transform=val_transform)

# Base loaders — train loader is built in the Loader Setup cell below
val_loader = DataLoader(
    val_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True,
    persistent_workers=NUM_WORKERS > 0,
)
test_loader = DataLoader(
    test_ds, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True,
    persistent_workers=NUM_WORKERS > 0,
)
print(f'Val  loader : {len(val_loader)} batches/epoch')
print(f'Test loader : {len(test_loader)} batches/epoch')
```

```python
train_labels     = train_df['level'].values
class_weights_np = compute_class_weight(
    class_weight='balanced', classes=np.arange(NUM_CLASSES), y=train_labels,
)
class_weights_t = torch.tensor(class_weights_np, dtype=torch.float32).to(device)

# Per-sample weights mirror the class weights — used by WeightedRandomSampler
sample_weights  = torch.as_tensor(class_weights_np[train_labels], dtype=torch.double)
weighted_sampler = WeightedRandomSampler(
    weights=sample_weights, num_samples=len(sample_weights), replacement=True,
)

print('Class weights (balanced inverse-freq):')
for name, w in zip(CLASS_NAMES, class_weights_np):
    print(f'  {name:<18}: {w:.4f}')
```

### FocalLoss

Focal loss down-weights easy, well-classified examples via the `(1 − pₜ)^γ` term. Used with `alpha=None` so WRS handles class balance at the batch level and FocalLoss focuses learning on hard examples.

```python
class FocalLoss(nn.Module):
    """Focuses learning on hard examples via (1 − pₜ)^γ weighting."""
    def __init__(self, gamma: float = 2.0):
        super().__init__()
        self.gamma = gamma

    def forward(self, logits, target):
        ce = F.cross_entropy(logits, target, reduction='none')
        pt = torch.exp(-ce)
        return ((1.0 - pt) ** self.gamma * ce).mean()


print('FocalLoss defined.')
```

## 6. Training Utilities

`run_epoch` — standard forward/backward pass.  
`run_epoch_clipped` — same but with optional gradient clipping (used in both training phases).  
Both functions operate on the global `model`, `criterion`, `optimizer`, `scaler` bindings.

```python
scaler = torch.amp.GradScaler('cuda', enabled=use_amp)


def run_epoch(loader, train_mode: bool):
    model.train(train_mode)
    total_loss, total = 0.0, 0
    all_preds, all_labels = [], []

    for images, labels in tqdm(loader, leave=False, desc='train' if train_mode else 'val'):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.autocast(device_type='cuda', enabled=use_amp):
            logits = model(images)
            loss   = criterion(logits, labels)

        if train_mode:
            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

        total_loss += loss.item() * labels.size(0)
        total      += labels.size(0)
        all_preds.extend(logits.argmax(dim=1).cpu().tolist())
        all_labels.extend(labels.cpu().tolist())

    return total_loss / total, all_labels, all_preds


def run_epoch_clipped(loader, train_mode: bool, grad_clip: float = 0.0):
    model.train(train_mode)
    total_loss, total = 0.0, 0
    all_preds, all_labels = [], []

    for images, labels in tqdm(loader, leave=False, desc='train' if train_mode else 'eval'):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        with torch.autocast(device_type='cuda', enabled=use_amp):
            logits = model(images)
            loss   = criterion(logits, labels)

        if train_mode:
            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            if grad_clip > 0:
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), grad_clip)
            scaler.step(optimizer)
            scaler.update()

        total_loss += loss.item() * labels.size(0)
        total      += labels.size(0)
        all_preds.extend(logits.argmax(dim=1).cpu().tolist())
        all_labels.extend(labels.cpu().tolist())

    return total_loss / total, all_labels, all_preds


print('run_epoch and run_epoch_clipped defined.')
```

## 7. Final Model — Two-Phase Fine-Tuning

Two-phase fine-tuning with **FocalLoss(α=None, γ=2) + WeightedRandomSampler** for class imbalance handling.

| Component | Choice |
|---|---|
| Loss | `FocalLoss(alpha=None, γ=2)` — hard-example focusing |
| Sampler | `WeightedRandomSampler` — balanced batch composition |
| Optional | Smart downsampling (`DOWNSAMPLING=True`) caps dominant classes |

**Training protocol:**  
Phase 1 → freeze backbone, train head only.  
Phase 2 → unfreeze all layers, fine-tune at a lower LR with gradient clipping.

### Loader Setup

Optionally downsamples the training set (each class capped at Grade-1 count). When downsampling is off, `WeightedRandomSampler` rebalances batch composition; when on, the dataset is already balanced so standard shuffle is used.

```python
# ── Optional smart downsampling — cap each class at Grade-1 count ────────────
if DOWNSAMPLING:
    _cap = int((train_df['level'] == 1).sum()) if DOWNSAMPLE_CAP is None else DOWNSAMPLE_CAP
    _parts = []
    for _g in range(NUM_CLASSES):
        _cls = train_df[train_df['level'] == _g]
        _parts.append(_cls.sample(n=min(len(_cls), _cap), random_state=SEED))
    _train_df = (pd.concat(_parts)
                 .sample(frac=1, random_state=SEED)
                 .reset_index(drop=True))
    print(f'Downsampling ON  : cap={_cap:,}  ({len(train_df):,} → {len(_train_df):,} samples)')
    for _g, _name in enumerate(CLASS_NAMES):
        _orig = (train_df['level'] == _g).sum()
        _new  = (_train_df['level'] == _g).sum()
        _note = '  ← capped' if _orig > _cap else ''
        print(f'  Grade {_g}  {_name:<18} {_orig:>6,} → {_new:>6,}{_note}')
else:
    _train_df = train_df
    _cap = None
    print(f'Downsampling OFF : using full training set  ({len(train_df):,} samples)')

_train_ds = DRDataset(_train_df, TRAIN_DIR, transform=train_transform)

# ── Build sampler matched to the actual training set size ────────────────────
# Always use WRS — when downsampling is on, recompute weights from the smaller set
# so indices stay within bounds (full-dataset sampler would go out of range).
_ds_labels  = _train_df['level'].values
_ds_cw      = compute_class_weight('balanced', classes=np.arange(NUM_CLASSES), y=_ds_labels)
_ds_sw      = torch.as_tensor(_ds_cw[_ds_labels], dtype=torch.double)
_train_sampler = WeightedRandomSampler(weights=_ds_sw, num_samples=len(_ds_sw), replacement=True)

train_loader = DataLoader(
    _train_ds, batch_size=BATCH_SIZE, sampler=_train_sampler,
    num_workers=NUM_WORKERS, pin_memory=True, drop_last=True,
    persistent_workers=NUM_WORKERS > 0,
)
_balance_method = ('Downsampling + FocalLoss(alpha=None, gamma=2) + WRS'
                   if DOWNSAMPLING else
                   'FocalLoss(alpha=None, gamma=2) + WeightedRandomSampler')

_sched_p1 = 'LinearWarmup+CosineDecay' if USE_WARMUP_P1 else 'CosineAnnealingLR'
_sched_p2 = 'LinearWarmup+CosineDecay' if USE_WARMUP_P2 else 'CosineAnnealingLR'
_warmup_note = f'  (warmup={WARMUP_EPOCHS} epochs)'

print()
print(f'Imbalance method  : {_balance_method}')
print(f'Phase 1 LR/Epochs : {PHASE1_LR:.0e} / {PHASE1_EPOCHS}')
print(f'Phase 2 LR/Epochs : {PHASE2_LR:.0e} / {PHASE2_EPOCHS}')
print(f'Batch / WD / Clip : {BATCH_SIZE} / {WEIGHT_DECAY} / {GRAD_CLIP}')
print(f'Patience          : {PATIENCE}')
print(f'Scheduler P1      : {_sched_p1}' + (_warmup_note if USE_WARMUP_P1 else ''))
print(f'Scheduler P2      : {_sched_p2}' + (_warmup_note if USE_WARMUP_P2 else ''))
print(f'Train loader      : {len(train_loader)} batches/epoch')
```

### Model, Criterion & Output Dirs

Fresh pretrained model, `FocalLoss(alpha=None, gamma=2)` as the fixed criterion, and checkpoint / figure directories created.

```python
set_seed(SEED)
vit_model  = timm.create_model(MODEL_NAME, pretrained=True, num_classes=NUM_CLASSES).to(device)
vit_scaler = torch.amp.GradScaler('cuda', enabled=use_amp)

vit_criterion = FocalLoss(gamma=2.0)  # alpha=None — WRS handles balance

FINAL_OUT_DIR  = ROOT / 'outputs' / 'vit_b16_final'
FINAL_CKPT_DIR = FINAL_OUT_DIR / 'checkpoints'
FINAL_FIG_DIR  = FINAL_OUT_DIR / 'figures'
for d in (FINAL_CKPT_DIR, FINAL_FIG_DIR):
    d.mkdir(parents=True, exist_ok=True)

PHASE1_CKPT = FINAL_CKPT_DIR / 'vit_b16_final_phase1_best.pt'
FINAL_CKPT        = FINAL_CKPT_DIR / 'vit_b16_final_best.pt'

print(f'Criterion  : {vit_criterion}')
print(f'Checkpoint → {FINAL_CKPT_DIR}')
```

### Phase 1 — Head Only (Frozen Backbone)

All layers except the classification head are frozen. Training only the head first prevents early distortion of the pre-trained features.

```python
# Freeze everything except the classification head
for name, param in vit_model.named_parameters():
    if 'head' not in name:
        param.requires_grad = False

trainable_p1 = sum(p.numel() for p in vit_model.parameters() if p.requires_grad)
print(f'Phase 1: {trainable_p1:,} trainable parameters (head only)\n')

# Stash any existing globals so we can restore them after training
_stash_model     = globals().get('model',     None)
_stash_criterion = globals().get('criterion', None)
_stash_optimizer = globals().get('optimizer', None)
_stash_scaler    = globals().get('scaler',    None)

model     = vit_model
criterion = vit_criterion
scaler    = vit_scaler
optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=PHASE1_LR, weight_decay=WEIGHT_DECAY,
)

if USE_WARMUP_P1:
    _p1_w = max(1, min(WARMUP_EPOCHS, PHASE1_EPOCHS - 1))
    scheduler_p1 = torch.optim.lr_scheduler.SequentialLR(optimizer, schedulers=[
        torch.optim.lr_scheduler.LinearLR(optimizer, start_factor=1e-2, end_factor=1.0, total_iters=_p1_w),
        torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=max(1, PHASE1_EPOCHS - _p1_w)),
    ], milestones=[_p1_w])
else:
    scheduler_p1 = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=PHASE1_EPOCHS)

vit_history = {
    'phase1_train': [], 'phase1_val': [], 'phase1_train_acc': [], 'phase1_val_acc': [], 'phase1_val_kappa': [],
    'phase2_train': [], 'phase2_val': [], 'phase2_train_acc': [], 'phase2_val_acc': [], 'phase2_val_kappa': [],
}
p1_best_val, p1_best_epoch, p1_patience = float('inf'), 0, PATIENCE

print(f"{'Epoch':<8} {'Tr Loss':>10} {'Tr Acc':>8} {'Val Loss':>10} {'Val Acc':>8} {'Val QWK':>9}")
print('─' * 60)

for epoch in range(1, PHASE1_EPOCHS + 1):
    tr_loss, tr_labels, tr_preds = run_epoch_clipped(train_loader, train_mode=True, grad_clip=0.0)
    vl_loss, vl_labels, vl_preds = run_epoch_clipped(val_loader, train_mode=False)
    scheduler_p1.step()

    tr_acc   = accuracy_score(tr_labels, tr_preds)
    vl_acc   = accuracy_score(vl_labels, vl_preds)
    vl_kappa = cohen_kappa_score(vl_labels, vl_preds, weights='quadratic')
    vit_history['phase1_train'].append(tr_loss)
    vit_history['phase1_val'].append(vl_loss)
    vit_history['phase1_train_acc'].append(tr_acc)
    vit_history['phase1_val_acc'].append(vl_acc)
    vit_history['phase1_val_kappa'].append(vl_kappa)

    marker = ''
    if vl_loss < p1_best_val:
        p1_best_val, p1_best_epoch, p1_patience = vl_loss, epoch, PATIENCE
        torch.save({'epoch': epoch, 'model_state': model.state_dict(), 'val_loss': vl_loss},
                   PHASE1_CKPT)
        marker = '  ← best'
    else:
        p1_patience -= 1

    print(f'{epoch:<8} {tr_loss:>10.4f} {tr_acc:>8.4f} {vl_loss:>10.4f} {vl_acc:>8.4f} {vl_kappa:>9.4f}{marker}')
    if p1_patience == 0:
        print(f'\nEarly stopping at epoch {epoch}.')
        break

print(f'\nPhase 1 done. Best val loss: {p1_best_val:.4f}  (epoch {p1_best_epoch})')
```

### Phase 2 — Full Fine-Tune (All Layers)

Reload Phase 1's best checkpoint, unfreeze everything, and fine-tune with a lower LR and gradient clipping.

```python
# Unfreeze everything and reload Phase 1's best weights
for param in vit_model.parameters():
    param.requires_grad = True

ckpt_p1 = torch.load(PHASE1_CKPT, map_location=device, weights_only=False)
model.load_state_dict(ckpt_p1['model_state'])
print(f'Phase 2: loaded Phase 1 best (epoch {ckpt_p1["epoch"]}, val_loss={ckpt_p1["val_loss"]:.4f})\n')

optimizer = torch.optim.AdamW(model.parameters(), lr=PHASE2_LR, weight_decay=WEIGHT_DECAY)

if USE_WARMUP_P2:
    _p2_w = max(1, min(WARMUP_EPOCHS, PHASE2_EPOCHS - 1))
    scheduler_p2 = torch.optim.lr_scheduler.SequentialLR(optimizer, schedulers=[
        torch.optim.lr_scheduler.LinearLR(optimizer, start_factor=1e-2, end_factor=1.0, total_iters=_p2_w),
        torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=max(1, PHASE2_EPOCHS - _p2_w)),
    ], milestones=[_p2_w])
else:
    scheduler_p2 = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=PHASE2_EPOCHS)

p2_best_val, p2_best_epoch, p2_patience = float('inf'), 0, PATIENCE

print(f"{'Epoch':<8} {'Tr Loss':>10} {'Tr Acc':>8} {'Val Loss':>10} {'Val Acc':>8} {'Val QWK':>9}")
print('─' * 60)

for epoch in range(1, PHASE2_EPOCHS + 1):
    tr_loss, tr_labels, tr_preds = run_epoch_clipped(train_loader, train_mode=True,  grad_clip=GRAD_CLIP)
    vl_loss, vl_labels, vl_preds = run_epoch_clipped(val_loader, train_mode=False)
    scheduler_p2.step()

    tr_acc   = accuracy_score(tr_labels, tr_preds)
    vl_acc   = accuracy_score(vl_labels, vl_preds)
    vl_kappa = cohen_kappa_score(vl_labels, vl_preds, weights='quadratic')
    vit_history['phase2_train'].append(tr_loss)
    vit_history['phase2_val'].append(vl_loss)
    vit_history['phase2_train_acc'].append(tr_acc)
    vit_history['phase2_val_acc'].append(vl_acc)
    vit_history['phase2_val_kappa'].append(vl_kappa)

    marker = ''
    if vl_loss < p2_best_val:
        p2_best_val, p2_best_epoch, p2_patience = vl_loss, epoch, PATIENCE
        torch.save({'epoch': epoch, 'model_state': model.state_dict(), 'val_loss': vl_loss},
                   FINAL_CKPT)
        marker = '  ← best'
    else:
        p2_patience -= 1

    print(f'{epoch:<8} {tr_loss:>10.4f} {tr_acc:>8.4f} {vl_loss:>10.4f} {vl_acc:>8.4f} {vl_kappa:>9.4f}{marker}')
    if p2_patience == 0:
        print(f'\nEarly stopping at epoch {epoch}.')
        break

print(f'\nPhase 2 done. Best val loss: {p2_best_val:.4f}  (epoch {p2_best_epoch})')
print(f'Checkpoint → {FINAL_CKPT}')

# Restore original globals
if _stash_model is not None:
    model, criterion, optimizer, scaler = (
        _stash_model, _stash_criterion, _stash_optimizer, _stash_scaler,
    )
```

```python
# Load best checkpoint into a dedicated model — avoids touching training globals
_xai_ckpt = ROOT / 'outputs' / 'vit_b16_final' / 'checkpoints' / 'vit_b16_final_best.pt'
xai_model  = timm.create_model(MODEL_NAME, pretrained=False, num_classes=NUM_CLASSES).to(device)
_ckpt      = torch.load(_xai_ckpt, map_location=device, weights_only=False)
xai_model.load_state_dict(_ckpt['model_state'])
xai_model.eval()
print(f'Loaded: {_xai_ckpt.name}  (epoch {_ckpt["epoch"]}, val_loss={_ckpt["val_loss"]:.4f})')

# Collect predictions over the full test set
xai_probs, xai_preds, xai_labels = [], [], []
with torch.no_grad():
    for imgs, labels in tqdm(test_loader, desc='Collecting predictions'):
        probs = F.softmax(xai_model(imgs.to(device)), dim=1).cpu().numpy()
        xai_probs.append(probs)
        xai_preds.append(probs.argmax(axis=1))
        xai_labels.append(labels.numpy())

xai_probs  = np.concatenate(xai_probs)
xai_preds  = np.concatenate(xai_preds)
xai_labels = np.concatenate(xai_labels)

print(f'Test samples : {len(xai_preds):,}')
print(f'Accuracy     : {(xai_preds == xai_labels).mean():.3f}')
print(f'QWK          : {cohen_kappa_score(xai_labels, xai_preds, weights="quadratic"):.4f}')

# Output directory for all XAI figures
XAI_DIR = ROOT / 'outputs' / 'vit_b16_final' / 'explainability'
XAI_DIR.mkdir(parents=True, exist_ok=True)

# One colour per DR grade
DR_COLORS = ['#2ecc71', '#f1c40f', '#e67e22', '#e74c3c', '#8e44ad']

# ── Image helpers ────────────────────────────────────────────────────────
def get_tensor(ds, idx):
    img, label = ds[idx]
    return img.unsqueeze(0).to(device), int(label)

def get_pil(ds, idx):
    # Undo ViT normalisation (mean=0.5, std=0.5) and return PIL image
    tensor, label = ds[idx]
    mean = torch.tensor(VIT_MEAN).view(3, 1, 1)
    std  = torch.tensor(VIT_STD).view(3, 1, 1)
    return TF.to_pil_image((tensor.cpu() * std + mean).clamp(0, 1)), int(label)

def to_np(pil):
    # PIL -> float32 numpy (H, W, 3) in [0, 1]
    arr = np.array(pil).astype(np.float32) / 255.0
    return (arr - arr.min()) / (arr.max() - arr.min() + 1e-8)

def blend(rgb_np, heatmap):
    # Resize heatmap, apply JET colormap, then use activation as per-pixel alpha
    # so low-activation regions stay transparent (show original) and hot regions
    # show the warm colormap — avoids the uniform blue-tint of a 50/50 linear mix.
    h, w    = rgb_np.shape[:2]
    heatmap = cv2.resize(heatmap, (w, h))
    heatmap = (heatmap - heatmap.min()) / (heatmap.max() - heatmap.min() + 1e-8)
    cmap    = cv2.applyColorMap(np.uint8(255 * heatmap), cv2.COLORMAP_JET)
    cmap    = cv2.cvtColor(cmap, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
    alpha   = heatmap[:, :, np.newaxis]          # (H, W, 1) activation as transparency
    out     = alpha * cmap + (1 - alpha) * rgb_np
    return np.uint8(255 * np.clip(out, 0, 1))

# Handy demo indices — first correct and first wrong prediction
correct_idx = int(np.where(xai_preds == xai_labels)[0][0])
wrong_idx   = int(np.where(xai_preds != xai_labels)[0][0])
print('Helpers ready')
```

```python
def get_gradcam(input_tensor, target_class):
    """
    Grad-CAM for ViT-B/16.

    Hooks model.blocks[-1] whose output is (B, 197, 768):
      index 0   = CLS token (no spatial location)
      index 1:  = 196 patch tokens on a 14x14 grid (224/16 = 14)

    Channel weights = global average pool of gradients.
    Weighted activation sum (no ReLU) is reshaped to 14x14 and returned
    as a min-max normalised float32 heatmap. ReLU is omitted because ViT
    patch activations are often all-negative, which zeros the entire map.
    """
    activations, gradients = {}, {}

    def fwd_hook(module, inp, out):
        activations['feat'] = out.detach()        # (1, 197, 768)

    def bwd_hook(module, grad_in, grad_out):
        gradients['feat'] = grad_out[0].detach()  # (1, 197, 768)

    h1 = xai_model.blocks[-1].register_forward_hook(fwd_hook)
    h2 = xai_model.blocks[-1].register_full_backward_hook(bwd_hook)

    xai_model.eval()
    out = xai_model(input_tensor)
    xai_model.zero_grad()
    out[0, target_class].backward()

    h1.remove(); h2.remove()

    # Patch tokens only (skip CLS at index 0), reshape to 14x14 spatial grid
    act  = activations['feat'][0, 1:]    # (196, 768)
    grad = gradients['feat'][0, 1:]      # (196, 768)
    w    = grad.mean(dim=0)              # (768,) global avg pool over patches
    cam  = (act * w).sum(-1).cpu().numpy().reshape(14, 14)  # (14, 14) raw scores
    # Suppress the bottom 30 % of patches (background noise) to sharpen contrast
    # around the truly attended regions before normalising to [0, 1].
    cam  = np.maximum(cam - np.percentile(cam, 30), 0)
    return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)

print('get_gradcam defined')
```

### 8.4 Attention Rollout — Native ViT Attention

Captures attention weights from all 12 transformer blocks and multiplies them
together (rollout) to trace how information flows from each patch to the CLS token.

Unlike Grad-CAM, no gradients are needed — it reads the model's own attention
mechanism directly, providing an independent cross-check of spatial focus.

```python
def attention_rollout(input_tensor):
    """
    ViT attention rollout — traces CLS token attention across all 12 blocks.

    timm's ViT uses fused attention (F.scaled_dot_product_attention) by default,
    which hides the attention matrix from hooks. We temporarily disable fused
    attention so attn_drop fires and the (B, heads, N, N) matrix is capturable,
    then immediately restore the original setting.

    Returns a normalised (14, 14) heatmap: which patches the CLS token
    attended to most, propagated through every layer.
    """
    # Expose attention weights by forcing the non-fused code path
    for blk in xai_model.blocks:
        blk.attn.fused_attn = False

    captured, hooks = {}, []

    def make_hook(k):
        def h(module, inp, out):
            captured[k] = out.detach().cpu()   # (B, heads, N, N)
        return h

    for i, blk in enumerate(xai_model.blocks):
        hooks.append(blk.attn.attn_drop.register_forward_hook(make_hook(i)))

    xai_model.eval()
    with torch.no_grad():
        _ = xai_model(input_tensor)

    for h in hooks: h.remove()
    # Restore fused attention for normal inference speed
    for blk in xai_model.blocks:
        blk.attn.fused_attn = True

    # Rollout: multiply attention matrices layer by layer
    # Add identity to account for residual connections, then renormalise rows
    rollout = None
    for k in range(len(xai_model.blocks)):
        attn    = captured[k][0].mean(0)                    # avg heads -> (197, 197)
        attn    = attn + torch.eye(attn.shape[-1])
        attn    = attn / attn.sum(-1, keepdim=True)
        rollout = attn if rollout is None else attn @ rollout

    # Row 0 = CLS token; columns 1: = patch tokens on a 14x14 grid
    mask = rollout[0, 1:].numpy().reshape(14, 14)
    return (mask - mask.min()) / (mask.max() - mask.min() + 1e-8)


def rollout_plot(ds, idx, save_path=None):
    # Three-panel: original | rollout overlay | raw attention map
    tensor, label = get_tensor(ds, idx)
    pil, _        = get_pil(ds, idx)
    rgb_np        = to_np(pil)

    with torch.no_grad():
        probs = F.softmax(xai_model(tensor), dim=1).cpu().numpy()[0]
    pred    = int(probs.argmax())
    amap    = attention_rollout(tensor)
    h, w    = rgb_np.shape[:2]
    amap_rs = cv2.resize(amap, (w, h))

    fig, axes = plt.subplots(1, 3, figsize=(14, 4))
    fig.suptitle(
        f'Attention Rollout  |  True: {CLASS_NAMES[label]}  ->  Pred: {CLASS_NAMES[pred]}'
        f'  ({probs[pred]*100:.1f}%)',
        fontsize=12, fontweight='bold',
        color='green' if pred == label else 'red',
    )
    axes[0].imshow(rgb_np)
    axes[0].set_title('Original'); axes[0].axis('off')

    axes[1].imshow(rgb_np)
    axes[1].imshow(amap_rs, alpha=0.55, cmap='inferno')
    axes[1].set_title('Rollout Overlay'); axes[1].axis('off')

    im = axes[2].imshow(amap_rs, cmap='inferno')
    axes[2].set_title('Attention Map'); axes[2].axis('off')
    plt.colorbar(im, ax=axes[2], fraction=0.046, pad=0.04)

    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()


# Demo on a Grade 2 correct prediction (most populated correct class)
g2_idx     = np.where(xai_labels == 2)[0]
g2_correct = g2_idx[xai_preds[g2_idx] == 2]
demo_idx   = int(g2_correct[0]) if len(g2_correct) else int(g2_idx[0])
rollout_plot(test_ds, demo_idx, XAI_DIR / 'attention_rollout.png')
```

### 8.5 Grade 1 False Negatives — Clinically Critical

**Grade 1 (Mild DR) predicted as Grade 0 (No DR)** is the most dangerous clinical
error: a patient with early, treatable disease is sent home without intervention.

Grad-CAM reveals what the model attended to when making the wrong call — typically
the bright optic disc rather than the peripheral microaneurysms that define Grade 1.

```python
fn_idx = np.where((xai_labels == 1) & (xai_preds == 0))[0]
print(f'Grade 1 -> Grade 0 false negatives: {len(fn_idx)}')

if len(fn_idx) == 0:
    print('No Grade 1 false negatives found.')
else:
    # Show up to 8 cases with the highest confidence in the wrong 'No DR' prediction
    worst  = fn_idx[np.argsort(xai_probs[fn_idx, 0])[::-1][:8]]
    n_show = min(8, len(worst))
    cols   = 4
    rows   = (n_show + cols - 1) // cols

    fig, axes = plt.subplots(rows * 2, cols, figsize=(cols * 4, rows * 7))
    axes = np.array(axes).reshape(rows * 2, cols)
    fig.suptitle(
        'Grade 1 -> Grade 0 False Negatives (Most Confident Errors)\n'
        'Top: original image  |  Bottom: Grad-CAM for the wrong No DR prediction',
        fontsize=11, fontweight='bold',
    )

    for i, idx in enumerate(worst[:n_show]):
        rb, col   = (i // cols) * 2, i % cols
        tensor, _ = get_tensor(test_ds, int(idx))
        pil, _    = get_pil(test_ds, int(idx))
        # Explain the wrong Grade 0 prediction to see what misled the model
        overlay   = blend(to_np(pil), get_gradcam(tensor, 0))
        conf_0    = xai_probs[idx, 0]

        axes[rb,     col].imshow(pil);     axes[rb,     col].axis('off')
        axes[rb,     col].set_title(f'True: Mild  Pred: No DR\nConf: {conf_0*100:.1f}%',
                                    fontsize=9, color='red')
        axes[rb + 1, col].imshow(overlay); axes[rb + 1, col].axis('off')
        axes[rb + 1, col].set_title('Grad-CAM (predicting No DR)', fontsize=8)

    for i in range(n_show, rows * cols):
        rb, col = (i // cols) * 2, i % cols
        axes[rb, col].axis('off'); axes[rb + 1, col].axis('off')

    plt.tight_layout()
    fig.savefig(XAI_DIR / 'grade1_false_negatives.png', dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Saved -> {XAI_DIR / "grade1_false_negatives.png"}')
```

### 8.6 LIME — Superpixel Explanations

LIME perturbs superpixel regions and observes how the prediction changes — it is
model-agnostic and provides an independent cross-check of Grad-CAM findings.

**Green regions** support the predicted class; **red regions** oppose it.
Comparing the two figures validates whether both spatial techniques agree on
the clinically relevant lesion areas.

```python
from lime import lime_image
from skimage.segmentation import mark_boundaries

lime_explainer = lime_image.LimeImageExplainer(random_state=42)


def lime_predict_fn(images_np):
    # Convert raw uint8 numpy arrays -> normalised tensors -> softmax probs
    batch = torch.stack(
        [val_transform(Image.fromarray(img)) for img in images_np]
    ).to(device)
    xai_model.eval()
    with torch.no_grad():
        return F.softmax(xai_model(batch), dim=1).cpu().numpy()


def lime_plot(ds, idx, num_samples=500, num_features=8, save_path=None):
    pil, label = get_pil(ds, idx)
    img_np     = np.array(pil.resize((224, 224)))   # LIME works on uint8 numpy

    with torch.no_grad():
        tensor, _ = get_tensor(ds, idx)
        probs     = F.softmax(xai_model(tensor), dim=1).cpu().numpy()[0]
    pred = int(probs.argmax())

    print(f'  Running LIME ({num_samples} samples)  '
          f'true={CLASS_NAMES[label]}  pred={CLASS_NAMES[pred]} ...')
    exp = lime_explainer.explain_instance(
        img_np, lime_predict_fn,
        top_labels=5, hide_color=0,
        num_samples=num_samples, random_seed=42,
    )

    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    fig.suptitle(
        f'LIME  |  True: {CLASS_NAMES[label]}  ->  Pred: {CLASS_NAMES[pred]}'
        f'  ({probs[pred]*100:.1f}%)',
        fontsize=12, fontweight='bold',
        color='green' if pred == label else 'red',
    )
    axes[0].imshow(img_np);  axes[0].set_title('Original');  axes[0].axis('off')

    # Regions that support the prediction (green boundary)
    pos, mask = exp.get_image_and_mask(
        pred, positive_only=True, num_features=num_features, hide_rest=False)
    axes[1].imshow(mark_boundaries(pos / 255.0, mask))
    axes[1].set_title('Supporting regions');  axes[1].axis('off')

    # All regions: green = support, red = oppose
    both, mask2 = exp.get_image_and_mask(
        pred, positive_only=False, num_features=num_features, hide_rest=False)
    axes[2].imshow(mark_boundaries(both / 255.0, mask2))
    axes[2].set_title('All regions (green=support, red=oppose)');  axes[2].axis('off')

    plt.tight_layout()
    if save_path:
        fig.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()


lime_plot(test_ds, correct_idx, save_path=XAI_DIR / 'lime_correct.png')
lime_plot(test_ds, wrong_idx,   save_path=XAI_DIR / 'lime_wrong.png')
```

### 8.7 Misclassification Analysis

Three complementary views of where the model fails:
- **Signed error histogram** — are errors biased toward under- or over-grading?
- **Row-normalised confusion matrix** — which pairs of grades are most confused?
- **Per-class Precision / Recall / F1** — which grade is hardest to detect?

```python
errors     = xai_preds.astype(int) - xai_labels.astype(int)
abs_errors = np.abs(errors)

# Signed error histogram + row-normalised confusion matrix
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].hist(errors, bins=np.arange(-4.5, 5.5), color='#3498db',
             edgecolor='white', rwidth=0.8)
axes[0].axvline(0, color='red', linestyle='--', lw=1.5, label='Correct')
axes[0].set_xlabel('Prediction Error (Predicted - True)')
axes[0].set_ylabel('Count')
axes[0].set_title('Signed Error Distribution', fontweight='bold')
axes[0].legend()

cm     = confusion_matrix(xai_labels, xai_preds)
cm_pct = cm.astype(float) / cm.sum(axis=1, keepdims=True) * 100
sns.heatmap(cm_pct, annot=True, fmt='.1f', cmap='Blues',
            xticklabels=[f'Pred G{i}' for i in range(5)],
            yticklabels=[f'True G{i}' for i in range(5)],
            ax=axes[1], cbar_kws={'label': '%'})
axes[1].set_title('Confusion Matrix (row-normalised %)', fontweight='bold')
axes[1].set_ylabel('True Grade'); axes[1].set_xlabel('Predicted Grade')

plt.suptitle('Misclassification Overview', fontsize=13, fontweight='bold')
plt.tight_layout()
fig.savefig(XAI_DIR / 'misclassification_overview.png', dpi=150, bbox_inches='tight')
plt.show()

for desc, count in [
    ('Correct        ', (abs_errors == 0).sum()),
    ('Off by 1 grade ', (abs_errors == 1).sum()),
    ('Off by >= 2    ', (abs_errors >= 2).sum()),
]:
    print(f'{desc}: {count:5d}  ({count/len(xai_labels)*100:.1f}%)')

# Worst severe misclassifications — Grad-CAM shows what the model fixated on
severe_idx = np.where(abs_errors >= 2)[0]
print(f'\nSamples with |error| >= 2: {len(severe_idx)}')

if len(severe_idx) >= 1:
    worst_severe = severe_idx[np.argsort(xai_probs[severe_idx, xai_preds[severe_idx]])[::-1][:4]]
    n_cols = min(4, len(worst_severe))
    fig2, axes2 = plt.subplots(2, n_cols, figsize=(n_cols * 4.5, 9))
    axes2 = np.array(axes2).reshape(2, n_cols)
    fig2.suptitle('Severe Misclassifications (|offset| >= 2) — Grad-CAM Analysis',
                  fontsize=12, fontweight='bold')
    for col, idx in enumerate(worst_severe[:n_cols]):
        tensor, label = get_tensor(test_ds, int(idx))
        pil, _        = get_pil(test_ds, int(idx))
        pred          = int(xai_preds[idx])
        conf          = xai_probs[idx, pred]
        overlay       = blend(to_np(pil), get_gradcam(tensor, pred))
        axes2[0, col].imshow(pil);     axes2[0, col].axis('off')
        axes2[0, col].set_title(f'True: {CLASS_NAMES[label]}\n'
                                f'Pred: {CLASS_NAMES[pred]}  ({conf*100:.1f}%)',
                                fontsize=9, color='red')
        axes2[1, col].imshow(overlay); axes2[1, col].axis('off')
        axes2[1, col].set_title('Grad-CAM', fontsize=8)
    plt.tight_layout()
    fig2.savefig(XAI_DIR / 'worst_misclassifications.png', dpi=150, bbox_inches='tight')
    plt.show()

# Per-class Precision, Recall, F1
f1_xai   = f1_score(xai_labels,        xai_preds, average=None, labels=list(range(5)))
prec_xai = precision_score(xai_labels, xai_preds, average=None, labels=list(range(5)), zero_division=0)
rec_xai  = recall_score(xai_labels,    xai_preds, average=None, labels=list(range(5)), zero_division=0)

x, w = np.arange(5), 0.22
fig3, ax = plt.subplots(figsize=(11, 4))
ax.bar(x - w, prec_xai, w, label='Precision', color='#3498db', edgecolor='white')
ax.bar(x,     rec_xai,  w, label='Recall',    color='#e67e22', edgecolor='white')
ax.bar(x + w, f1_xai,   w, label='F1',        color='#2ecc71', edgecolor='white')
ax.set_xticks(x)
ax.set_xticklabels([f'G{i} {CLASS_NAMES[i]}' for i in range(5)])
ax.set_ylim(0, 1.15); ax.set_ylabel('Score')
ax.set_title('Per-Class Precision, Recall, F1 — Test Set', fontweight='bold')
ax.legend(); ax.grid(axis='y', alpha=0.3)
for i, (p, r, f) in enumerate(zip(prec_xai, rec_xai, f1_xai)):
    ax.text(i - w, p + 0.02, f'{p:.2f}', ha='center', fontsize=7)
    ax.text(i,     r + 0.02, f'{r:.2f}', ha='center', fontsize=7)
    ax.text(i + w, f + 0.02, f'{f:.2f}', ha='center', fontsize=7)
plt.tight_layout()
fig3.savefig(XAI_DIR / 'per_class_metrics.png', dpi=150, bbox_inches='tight')
plt.show()
```

### 8.8 Confidence Calibration

A well-calibrated model's stated confidence should match real-world accuracy:
if it says 70% on 100 images, ~70 should be correct.

- **Reliability diagrams** — one per grade; ECE (Expected Calibration Error) in title
- **Confidence vs accuracy tradeoff** — at threshold T, how accurate are retained
  predictions and what fraction of the test set is kept?

```python
# Per-grade reliability diagrams
fig, axes = plt.subplots(1, 5, figsize=(17, 4), sharey=True)
fig.suptitle('Reliability Diagrams — Confidence Calibration per DR Grade',
             fontsize=12, fontweight='bold')

for grade in range(5):
    y_bin  = (xai_labels == grade).astype(int)
    y_prob = xai_probs[:, grade]
    try:
        frac, mean_p = calibration_curve(y_bin, y_prob, n_bins=10, strategy='quantile')
    except ValueError:
        frac, mean_p = np.array([0, 1]), np.array([0, 1])

    # Expected Calibration Error: weighted mean of |confidence - accuracy| per bin
    ece = sum(
        m.sum() / len(y_prob) * abs(y_bin[m].mean() - y_prob[m].mean())
        for lo, hi in zip(np.linspace(0, 1, 11)[:-1], np.linspace(0, 1, 11)[1:])
        if (m := (y_prob >= lo) & (y_prob < hi)).sum() > 0
    )

    ax = axes[grade]
    ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Perfect')
    ax.plot(mean_p, frac, 'o-', color=DR_COLORS[grade], lw=2, ms=5, label='Model')
    ax.fill_between(mean_p, frac, mean_p, alpha=0.12, color=DR_COLORS[grade])
    ax.set_title(f'Grade {grade}\nECE={ece:.3f}', fontsize=10,
                 fontweight='bold', color=DR_COLORS[grade])
    ax.set_xlabel('Mean Predicted Prob.')
    if grade == 0: ax.set_ylabel('Fraction Positive')
    ax.set_xlim(0, 1); ax.set_ylim(0, 1)
    ax.legend(fontsize=7); ax.grid(alpha=0.3)

plt.tight_layout()
fig.savefig(XAI_DIR / 'calibration_reliability.png', dpi=150, bbox_inches='tight')
plt.show()

# Confidence vs accuracy tradeoff
top_conf = xai_probs.max(axis=1)
correct  = (xai_preds == xai_labels).astype(int)
thresh   = np.linspace(0.0, 0.95, 40)
accs, covs = [], []
for t in thresh:
    m = top_conf >= t
    if m.sum() == 0: break
    accs.append(correct[m].mean())
    covs.append(m.mean())

fig2, ax1 = plt.subplots(figsize=(9, 4))
ax2 = ax1.twinx()
ax1.plot(thresh[:len(accs)], accs, 'b-o',  ms=4, lw=2, label='Accuracy above threshold')
ax2.plot(thresh[:len(accs)], covs, 'r--s', ms=4, lw=2, label='Coverage (fraction retained)')
ax1.set_xlabel('Confidence Threshold')
ax1.set_ylabel('Accuracy', color='blue')
ax2.set_ylabel('Coverage', color='red')
ax1.set_ylim(0, 1); ax2.set_ylim(0, 1)
ax1.set_title('Confidence vs Accuracy Tradeoff', fontweight='bold')
lines1, lbl1 = ax1.get_legend_handles_labels()
lines2, lbl2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, lbl1 + lbl2)
ax1.grid(alpha=0.3)
plt.tight_layout()
fig2.savefig(XAI_DIR / 'confidence_accuracy_tradeoff.png', dpi=150, bbox_inches='tight')
plt.show()
```
